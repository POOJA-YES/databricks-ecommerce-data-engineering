```python
from pyspark.sql import functions as F
inventory_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(f"{DATA_PATH}/inventory.csv")
)

inventory_bronze = (
    inventory_df
    .withColumn("_ingestion_timestamp", F.current_timestamp())
    .withColumn("_source_file", F.lit("inventory.csv"))
)

(
    inventory_bronze.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.bronze.inventory")
)

display(inventory_bronze)
```
