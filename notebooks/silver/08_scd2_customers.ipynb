```python
from pyspark.sql import functions as F

customers = spark.table("workspace.silver.customers")

scd2_customers = (
    customers
    .withColumn("effective_start_date", F.col("updated_at"))
    .withColumn(
        "effective_end_date",
        F.to_timestamp(F.lit("9999-12-31 23:59:59"))
    )
    .withColumn("is_current", F.lit(True))
)

(
    scd2_customers.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.silver.customers_scd2")
)

display(scd2_customers)
```

```python
%sql
SELECT *
FROM workspace.silver.customers_scd2
ORDER BY customer_id;
```
