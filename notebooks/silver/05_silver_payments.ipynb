```python
from pyspark.sql import functions as F

df = spark.table("workspace.bronze.payments")

silver_payments = (
    df
    .withColumn("payment_date", F.to_timestamp("payment_date", 'dd-MM-yyyy HH:mm'))
    .withColumn("amount", F.col("amount").cast("decimal(12,2)"))
    .withColumn("payment_status", F.upper(F.trim("payment_status")))
    .dropDuplicates(["payment_id"])
    .filter(F.col("amount") >= 0)
)

(
    silver_payments.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.silver.payments")
)

```
