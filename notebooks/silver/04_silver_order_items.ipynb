```python
from pyspark.sql import functions as F

df = spark.table("workspace.bronze.order_items")

silver_order_items = (
    df
    .withColumn("quantity", F.col("quantity").cast("int"))
    .withColumn("unit_price", F.col("unit_price").cast("decimal(12,2)"))
    .withColumn("discount", F.col("discount").cast("decimal(12,2)"))
    .withColumn("tax", F.col("tax").cast("decimal(12,2)"))
    .withColumn("line_total", F.col("line_total").cast("decimal(12,2)"))
    .filter(F.col("quantity") > 0)
    .filter(F.col("order_id").isNotNull())
    .filter(F.col("product_id").isNotNull())
    .dropDuplicates(["order_item_id"])
)

(
    silver_order_items.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("workspace.silver.order_items")
)

```
